Here we download the CSV file containing the text snippets and author IDs.

```python
df = pd.read_csv('Gungor_2018_VictorianAuthorAttribution_data-train.csv', encoding = "ISO-8859-1")
df.head()
```

```python
text = list(df['text'])
labels = df['author'].values
```

### Exercises

1. Use the `CountVectorizer` from `sklearn.feature_extraction.text` to produce a term frequency vector for each text.  Set `max_features=1000` to only use the top 1000 terms.

Prepare a 90/10 train-test split `random_state=42`.

Train the default `MLPCLassifier` from `sklearn.neural_network` on the data and report the train and test accuracy.  You can use the argument `verbose=True` to `MLPClassifier` to monitor training.

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(max_features=1000)
X = vectorizer.fit_transform(text)
```

```python
vectorizer.get_feature_names_out()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, labels, train_size=0.9, random_state=42)

# change to np array from scipy sparse matrix type
X_train = X_train.toarray() 
X_test = X_test.toarray()

print(f'X_train shape: {X_train.shape} of type {type(X_train)}')
print(f'X_test shape: {X_test.shape} of type {type(X_test)}')
print(f'y_train shape: {y_train.shape} of type {type(y_train)}')
print(f'y_test shape: {y_test.shape} of type {type(y_test)}')
```

```python
from sklearn.neural_network import MLPClassifier

model = MLPClassifier(verbose=True).fit(X_train, y_train)
```

```python
print(f'The final train accuracy is: {model.score(X_train, y_train)}')
print(f'The final test accuracy is:  {model.score(X_test, y_test)}')
```


2. Repeat the steps but using `TfidfVectorizer` to produce term frequency - inverse document frequency vectors.

Does the IDF weighting improve the results?

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer_tfidf = TfidfVectorizer(max_features=1000)
X_tfidf = vectorizer_tfidf.fit_transform(text)
```

```python
X_train_tfidf, X_test_tfidf, y_train_tfidf, y_test_tfidf = train_test_split(X_tfidf, labels, train_size=0.9, random_state=42)

# change to np array from scipy sparse matrix type
X_train_tfidf = X_train_tfidf.toarray() 
X_test_tfidf = X_test_tfidf.toarray()

print(f'X_train shape: {X_train_tfidf.shape} of type {type(X_train_tfidf)}')
print(f'X_test shape: {X_test_tfidf.shape} of type {type(X_test_tfidf)}')
print(f'y_train shape: {y_train_tfidf.shape} of type {type(y_train_tfidf)}')
print(f'y_test shape: {y_test_tfidf.shape} of type {type(y_test_tfidf)}')
```

```python
model_tfidf = MLPClassifier(verbose=True).fit(X_train_tfidf, y_train_tfidf)
```

```python
print(f'The final train accuracy with tfidf is: {model_tfidf.score(X_train_tfidf, y_train_tfidf)}')
print(f'The final test accuracy with tfidf is:  {model_tfidf.score(X_test_tfidf, y_test_tfidf)}')
```

Overall, it seems that the tfidf in this case did not increase test accuracy, in fact it was slightly lower by about 0.6%. The train accuracies for both tokenizers were essentially 100%.
